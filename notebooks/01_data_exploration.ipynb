### Первичный анализ данных

На первом этапе была изучена структура датасета Online Retail. Исходный набор содержит 541 909 строк и 8 столбцов. Данные включают номера счетов, информацию о товарах, количество, цену, дату операции, идентификаторы клиентов и страну.

В ходе проверки были выявлены следующие особенности:

- В столбце `Description` отсутствует 1 454 значения.
- В столбце `CustomerID` отсутствует 135 080 значений, что составляет примерно 24,93% строк.
- Обнаружено 5 268 полных дубликатов строк.
- В столбце `Quantity` найдено 10 624 отрицательных значения.
- В столбце `UnitPrice` найдено 2 отрицательных значения.
- В 9 288 строках номер счёта начинается с буквы `C`. Все эти строки имеют отрицательное количество.
- Ещё 1 336 строк имеют отрицательное количество, но номер счёта не начинается с `C`. Все эти строки также имеют нулевую цену.

Отрицательные количества, нулевые и отрицательные цены, пропуски и дубликаты требуют дополнительного изучения. До завершения анализа эти записи не удаляются, поскольку они могут отражать возвраты, отмены или другие особенности учёта операций.

```python
print(df[(df['Quantity'] < 0) & (df['UnitPrice'] == 0)]['Description'].isnull().sum())
```

- Среди 1 336 строк с отрицательным количеством и нулевой ценой у 862 строк (примерно 64,52%) отсутствует описание товара. Это затрудняет определение содержания операций, поэтому данную группу необходимо исследовать отдельно.

```python
print(df[(df['Quantity'] < 0) & (df['UnitPrice'] == 0)]['CustomerID'].isnull().sum())
```

- Среди 1336 строк с отрицательным количеством и нулевой ценой у 1336 строк отсутствует CustomerID. Отсутствие идентификатора клиента затрудняет анализ, мы не можем определить количество и долю таких операций для конкретного клиента, поскольку нет CustomerID

```python
import pandas as pd
data_cleaning_rules = pd.DataFrame(columns=['issue', 'count', 'decision', 'reason', 'impact']) #создание таблицы
print(data_cleaning_rules.shape) 
```

```python
data_cleaning_rules.loc[0] = ['Полные дубликаты строк', 5268, 'Удалить повторяющиеся полные строки, оставив по одной копии каждой уникальной записи', 'Повторяющиеся записи могут привести к двойному учёту операций и исказить результаты анализа', 'После удаления количество строк уменьшится, а рассчитанная сумма продаж может измениться']
```

```python
print(data_cleaning_rules.dtypes) #тип данных
```

```python
data_cleaning_rules.loc[1] = ['Пропущенные значения в Description', 1454, 'Оставить строки, если остальные данные полезны для анализа', 'Описание отсутствует, но код товара и другие данные могут быть заполнены', 'Анализ по названиям товаров будет неполным, но остальные расчёты можно сохранить']
```

```python
data_cleaning_rules.loc[2] = ['Пропущенные значения в CustomerID', 135080, 'оставить строки, если остальные данные полезны для анализа', 'для расчёта выручки идентификатор покупателя не нужен, поскольку выручка рассчитывается по количеству товара и цене', 'без идентификатора нельзя надёжно связать такие операции с конкретными покупателями, поэтому анализ клиентских покупок будет неполным.']
```

```python
data_cleaning_rules.loc[3] = ['Отрицательные значения в Quantity', 10624, 'сохранить строки в исходных данных, а отмены и другие необычные операции анализировать отдельно от обычных продаж', 'отрицательное количество может обозначать отмену или возврат, но назначение всех записей не подтверждено. Поэтому удалять их без дополнительной проверки нельзя', 'отрицательные количества могут уменьшать рассчитанную выручку, а смешивание отмен и обычных продаж затрудняет анализ продаж и возвратов']
```

```python
data_cleaning_rules.loc[4] = ['Отрицательные значения в UnitPrice', 2, 'сохранить записи в исходных данных и исключить их из расчёта выручки от обычных продаж до выяснения их назначения. Так мы не потеряем информацию', 'назначение записей требует дополнительной проверки. В обеих указано Adjust bad debt, что указывает на бухгалтерскую корректировку, а не на обычную продажу', 'если включить отрицательные цены в общую сумму выручки от продаж, результат может оказаться заниженным. Поэтому такие записи следует анализировать отдельно']
```

```python
# Создаём копию исходных данных
df_clean = df.copy()

# Удаляем полные дубликаты
df_clean = df_clean.drop_duplicates()

# Проверяем размер таблицы
print(df_clean.shape)
```

```python
df_clean['Revenue'] = df_clean['Quantity'] * df_clean['UnitPrice'] #добавление столбца
print(df_clean.head())
```

```python
print(df_clean.nunique())
```

```python
#это предварительная выборка обычных продаж
sales = df_clean[
    (df_clean['InvoiceNo'].str[0] != 'C') & 
    (df_clean['UnitPrice'] > 0) & 
    (df_clean['Quantity'] > 0)]
```

```python
print(sales.shape[0])
```

```python
#Общая выручка обычных продаж
print(sales['Revenue'].sum())
```

```python
#группировка по стране, сумма выручки, сортировка по убыванию топ про странам
sales.groupby('Country')['Revenue'].sum().sort_values(ascending=False).head(10)
```

```python
##группировка по месяцу , сумма выручки, сортировка по убыванию топ по месяцам
sales.groupby(sales['InvoiceDate'].dt.to_period('M'))['Revenue'].sum().sort_values(ascending=False).head(10)
```

```python
# топ по товарам
sales.groupby('StockCode')['Revenue'].sum().sort_values(ascending=False).head(10)
```

```python
#Отдельно исследую отмены и возвраты
cancellations = df_clean[(df_clean['InvoiceNo'].str[0] == 'C')][['StockCode', 'Description', 'Quantity', 'UnitPrice', 'Revenue']]
print(cancellations.head(10))
```
